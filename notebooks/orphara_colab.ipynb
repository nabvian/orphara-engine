## 4. Analyze your own note
Edit the text below, run the cell to save it, then run the analysis. You can add report files too (upload them with the Colab file browser on the left, then list their names after `--`).

```python
note = """Clinical Note
Date: 4 May 2026
28-year-old man with jaundice and dark urine.
Blood film shows bite cells and Heinz bodies.
Haemolysis following oxidant exposure.
G6PD assay low, confirming the finding.
Spleen not enlarged.
"""
with open('my_note.txt', 'w') as f:
    f.write(note)
print('saved my_note.txt')
```

```python
!cargo run -q -p concept-adapter --example analyze_note -- my_note.txt
```

## 5. What you are seeing

The engine is **disease-blind** — it knows nothing about any specific disease. It reasons over whatever compiled model it is given; here that is the bundled synthetic example. A value it cannot map is *quarantined* (shown, never invented); a candidate whose confirmatory concept is missing is only *Possible*, and the engine names what is missing.

To run the same engine over a different model, set `ORPHARA_MODEL` to a `.dsdsrc` path before the command, e.g. `!ORPHARA_MODEL=my_model.dsdsrc cargo run -q -p concept-adapter --example analyze_note -- my_note.txt`.

The real disease knowledge base (a proof set of three red-cell enzyme disorders, being built out over the coming year across the rare-disease Hematology family, and ultimately the whole Orphanet inventory) is kept private until clinical validation, after which the full suite will be open-sourced.
